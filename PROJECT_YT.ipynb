```python
# USES PANDAS AND NUMPY
import pandas as pd #used for data analysis
import numpy as np #used for numerical python

# loading the dataset
df = pd.read_csv("project_data.csv")
print(df.head())
```

```python
# checking the missing values
print('Missing values in each columns')
print(df.isnull().sum())
```

```python
import numpy as np

# Fill missing Salary with mean
df['Salary (INR)'] = df['Salary (INR)'].fillna(df['Salary (INR)'].mean())

# Fill missing Performance Rating with median
df['Performance Rating'] = df['Performance Rating'].fillna(
    df['Performance Rating'].median()
)

# Replace inf and -inf with NaN
df = df.replace([np.inf, -np.inf], np.nan)

# Fill remaining numeric NaN values with mean
df = df.fillna(df.mean(numeric_only=True))

# Remove duplicate records
df = df.drop_duplicates()

# Replace negative salaries
salary_mean = df['Salary (INR)'].mean()
df['Salary (INR)'] = np.where(
    df['Salary (INR)'] < 0,
    salary_mean,
    df['Salary (INR)']
)

# Handle unrealistic salary values (3-sigma rule)
salary_std = df['Salary (INR)'].std()
lower_bound = salary_mean - (3 * salary_std)
upper_bound = salary_mean + (3 * salary_std)

df = df[
    (df['Salary (INR)'] >= lower_bound) &
    (df['Salary (INR)'] <= upper_bound)
]

# Save cleaned data
df.to_csv('cleaned_project_data.csv', index=False)

print('DATA CLEANING COMPLETED! SAVED AS "cleaned_project_data.csv"')
```
